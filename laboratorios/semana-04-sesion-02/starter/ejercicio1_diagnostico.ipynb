# Ejercicio 1: diagnóstico completo de un archivo de sensores

`lecturas_sensores.csv`: 10 lecturas de temperatura, con filas repetidas, una celda vacía, un `N/D` y fechas en dos formatos.

Corre todas las celdas antes de tocar nada (**Run All**, en la barra de arriba del notebook): corre de principio a fin sin los `TODO` resueltos. Fíjate en qué números y qué fechas salen así. Después resuelve cada `TODO` en su celda y vuelve a correr desde esa celda hacia abajo.

```python
import pandas as pd

datos = pd.read_csv("lecturas_sensores.csv")
print(f"Filas leidas: {len(datos)}")
print(datos)
```

## Paso 1: duplicados

```python
# TODO 1: cuenta las filas que repiten todas sus columnas y despues
# quitalas de datos.
duplicados = int(datos.duplicated().sum())
datos = datos.drop_duplicates()
print(f"Filas duplicadas: {duplicados}")
print(f"Filas despues de quitarlas: {len(datos)}")
```

## Paso 2: tipos y faltantes visibles

```python
print(datos.dtypes)
# TODO 2: cuenta los valores faltantes de cada columna.
faltantes = datos.isna().sum()
print(faltantes)
```

## Paso 3: faltantes disfrazados de texto

```python
# TODO 3: convierte temperatura a numero. Lo que no se pueda convertir
# tiene que quedar como faltante, no detener el notebook.
datos["temperatura_num"] = pd.to_numeric(datos["temperatura"], errors="coerce")
print(f"Temperaturas faltantes antes de convertir: {datos['temperatura'].isna().sum()}")
print(f"Temperaturas faltantes despues de convertir: {datos['temperatura_num'].isna().sum()}")
```

## Paso 4: fechas con dos formatos

```python
def normalizar_fecha(valor):
    # TODO 4: si valor tiene "/", viene como dia/mes/anio; si no, como
    # anio-mes-dia. Convierte cada caso con su formato explicito.
    if "/" in valor:
        return pd.to_datetime(valor, format="%d/%m/%Y")
    return pd.to_datetime(valor, format="%Y-%m-%d")


datos["fecha_normalizada"] = datos["fecha"].apply(normalizar_fecha)
datos[["sensor_id", "fecha", "fecha_normalizada"]]
```

## Resultado

```python
datos.to_csv("lecturas_limpias.csv", index=False)
print(f"Guardado lecturas_limpias.csv con {len(datos)} filas.")
```
